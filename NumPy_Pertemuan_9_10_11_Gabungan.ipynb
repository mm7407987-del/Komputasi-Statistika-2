## 2. Creating NumPy Arrays

### 2.1 Creating Arrays from Python Lists

Fungsi `np.array()` digunakan untuk mengubah data Python menjadi NumPy Array. Jumlah dimensi ditentukan oleh struktur data yang dimasukkan.
- **Array 1D:** Berbentuk vektor.
- **Array 2D:** Berbentuk matriks (baris dan kolom).
- **Array 3D:** Kumpulan matriks 2D.
- **Array 4D:** Kumpulan array 3D.

Harus memiliki bentuk (`shape`) yang konsisten. **Shape:** Ukuran NumPy Array yang menunjukkan jumlah elemen pada setiap dimensi.

### 2.2 Creating a One-Dimensional Array (1D)

A 1D array is useful for representing measurements of a **single variable**, such as the scores of five students.

````python
student_scores = [75, 80, 92, 86, 88]
scores_1d = np.array(student_scores)

print(scores_1d)
print("Dimensions:", scores_1d.ndim)
print("Shape:", scores_1d.shape)
````

### 2.3 Creating a Two-Dimensional Array (2D)

A 2D array represents data **in rows and columns**. In the example below, each row corresponds to a student, while each column corresponds to a different assessment.

````python
student_assessments = [[80, 85, 90],
                       [75, 88, 84],
                       [92, 90, 95]]

scores_2d = np.array(student_assessments)

print(scores_2d)
print("Dimensions:", scores_2d.ndim)
print("Shape:", scores_2d.shape)
````

The shape `(3, 3)` indicates **three rows and three columns**.

### 2.4 Creating a Three-Dimensional Array (3D)

A 3D array can be understood as **a stack of matrices**.

````python
week_1 = np.array([[10, 12],
                   [14, 16]])
week_2 = np.array([[11, 13],
                   [15, 17]])

observations_3d = np.array([week_1, week_2])

print(observations_3d)
print("Dimensions:", observations_3d.ndim)
print("Shape:", observations_3d.shape)
````

The shape `(2, 2, 2)` indicates **two matrices**, each containing **two rows and two columns**. The first axis identifies which matrix is selected.

### 2.5 Creating a Four-Dimensional Array (4D)

A 4D array adds another organizational level. You can think of it as a collection of 3D arrays. Such structures appear in scientific data and batches of multidimensional observations.

````python
first_cube = np.array([[[1, 2], [3, 4]],
                       [[5, 6], [7, 8]]])
second_cube = np.array([[[11, 12], [13, 14]],
                        [[15, 16], [17, 18]]])

observations_4d = np.array([first_cube, second_cube])

print(observations_4d)
print("Dimensions:", observations_4d.ndim)
print("Shape:", observations_4d.shape)
````

### 2.6 Creating Arrays from Scratch: `np.zeros()`
`np.zeros()` directly creates an array filled with zeros.

````python
zero_vector = np.zeros(5)
zero_matrix = np.zeros((3, 4))

print("1D zeros:")
print(zero_vector)
print("2D zeros:")
print(zero_matrix)
````

### 2.7 Creating Random Arrays: `np.random.random()`

`np.random.random()` menghasilkan bilangan desimal acak antara **0 hingga kurang dari 1**. Penggunaan `seed` membuat hasil bilangan acak tetap sama setiap kali kode dijalankan.

````python
np.random.seed(42)
random_measurements = np.random.random((2, 3))

print(random_measurements)
print("Shape:", random_measurements.shape)
````

### 2.8 Creating Sequences: `np.arange()`

`np.arange(start, stop, step)` digunakan untuk membuat deret angka dengan nilai awal (`start`), batas akhir (`stop`), dan jarak antarangka (`step`). Nilai `stop` tidak disertakan. Secara default, `start = 0` dan `step = 1`.

````python
sequence_a = np.arange(5)
sequence_b = np.arange(2, 11, 2)
sequence_c = np.arange(-3, 4, 3)

print("Default sequence:", sequence_a)
print("Even-number sequence:", sequence_b)
print("Stepped sequence:", sequence_c)
````

## 3. Array Attributes and Data Types

### 3.1 Array Dimensionality and Axes

**Axis** adalah sumbu yang menunjukkan arah dimensi pada NumPy Array. Pada array 2D, `axis=0` menunjukkan arah baris dan `axis=1` menunjukkan arah kolom. Pada array 3D, terdapat tiga sumbu, yaitu lapisan, baris, dan kolom.

- **`.ndim`**: Jumlah dimensi atau sumbu.
- **`.shape`**: Ukuran setiap dimensi.
- **`.size`**: Jumlah seluruh elemen array.

| Attribute | Meaning | Example result |
|---|---|---|
| `.ndim` | Number of dimensions/axes | `2` |
| `.shape` | Length along each axis | `(3, 4)` |
| `.size` | Total number of elements | `12` |
| `.dtype` | Common data type of the elements | `int64`, `float64`, etc. |

````python
measurement_matrix = np.array([[12, 15, 18, 21],
                               [14, 16, 20, 22],
                               [11, 13, 17, 19]])

print("Array:")
print(measurement_matrix)
print("ndim:", measurement_matrix.ndim)
print("shape:", measurement_matrix.shape)
print("size:", measurement_matrix.size)
print("dtype:", measurement_matrix.dtype)
````

### 3.2 Understanding Shape and Size

A shape of `(3, 4)` means the array has three rows and four columns; its size is therefore `3 × 4 = 12` elements. For a 3D or 4D array, the total size is the product of all entries in its shape.

````python
print("1D:", scores_1d.ndim, scores_1d.shape, scores_1d.size)
print("2D:", scores_2d.ndim, scores_2d.shape, scores_2d.size)
print("3D:", observations_3d.ndim, observations_3d.shape, observations_3d.size)
print("4D:", observations_4d.ndim, observations_4d.shape, observations_4d.size)
````

### 3.3 Common NumPy Data Types (`dtype`)

NumPy includes types for integers (`int`), floating-point numbers (`float`), Boolean values (`bool`), and strings (`str`). The actual `dtype` shown by NumPy provides more detail, such as the number of bits used for a numeric value.

Because NumPy arrays use a common data type, combining different kinds of values may cause automatic type conversion.

````python
integer_values = np.array([1, 2, 3])
float_values = np.array([1.5, 2.5, 3.5])
boolean_values = np.array([True, False, True])
string_values = np.array(["low", "medium", "high"])

print("Integer:", integer_values.dtype)
print("Float:", float_values.dtype)
print("Boolean:", boolean_values.dtype)
print("String:", string_values.dtype)
````

### 3.4 Specifying a Data Type Explicitly

Use the `dtype` argument in `np.array()` to request a specific type. A smaller numeric type may occupy less memory, but it also has a more limited range or precision.

````python
measurements_float32 = np.array([1.25, 2.75, 3.50], dtype=np.float32)
counts_int32 = np.array([10, 20, 30], dtype=np.int32)

print(measurements_float32, measurements_float32.dtype)
print(counts_int32, counts_int32.dtype)
````

````python
measurements_int32 = np.array([1.25, 2.75, 3.50], dtype=np.int32)
measurements_int32 
````

### 3.5 Type Conversion with `astype()`

The `.astype()` method returns an array converted to the specified data type. A common application is converting strings that contain valid numbers into numeric values.

Be careful: converting a floating-point number to an integer discards its fractional part; it is not the same as rounding to the nearest integer.

````python
text_scores = np.array(["70", "85", "90"])
numeric_scores = text_scores.astype(np.int64)

print("Before:", text_scores, text_scores.dtype)
print("After:", numeric_scores, numeric_scores.dtype)
````

````python
continuous_values = np.array([1.9, 2.1, 3.8])
integer_conversion = continuous_values.astype(int)

print("Original:", continuous_values)
print("Converted:", integer_conversion)
````

### 3.6 Type Coercion

**Type coercion** occurs when NumPy automatically chooses a common data type for the elements used to create an array. Examples include converting integers to floats when a decimal is included, or converting numeric values to strings when they are mixed with text.

**Always inspect `.dtype` before numerical analysis.**

````python
coercion_a = np.array([10, 20.5, 30])
coercion_b = np.array([True, False, 5])
coercion_c = np.array([1, "two", 3])

print(coercion_a, coercion_a.dtype)
print(coercion_b, coercion_b.dtype)
print(coercion_c, coercion_c.dtype)
````

## 4. Reshaping Arrays

### 4.1 Understanding Array Shape

The **shape** describes how the elements are organized. Importantly, a 1D array with six elements has shape `(6,)`, not `(1, 6)`. A row-shaped 2D array and a column-shaped 2D array have different shapes, even when they contain the same six values.

````python
base_values = np.arange(1, 7)

row_array = base_values.reshape(1, 6)
column_array = base_values.reshape(6, 1)

print("1D shape:", base_values.shape)
print("Row shape:", row_array.shape)
print(row_array)
print("Column shape:", column_array.shape)
print(column_array)
````

### 4.2 Reshaping Arrays with `reshape()`

`.reshape()` changes the arrangement of the elements without changing the total number of elements. A 1D array containing 12 values can be reshaped into `(3, 4)`, `(4, 3)`, or `(2, 2, 3)` because each shape contains exactly 12 positions.

For this basic example, NumPy reads the elements in their existing order and organizes them into the new shape.

````python
values_12 = np.arange(1, 13)

matrix_3x4 = values_12.reshape(3, 4)
matrix_4x3 = values_12.reshape(4, 3)
array_2x2x3 = values_12.reshape(2, 2, 3)

print("Original:", values_12)
print("3 × 4:")
print(matrix_3x4)
print("4 × 3:")
print(matrix_4x3)
print("2 × 2 × 3:")
print(array_2x2x3)
````

### 4.3 Reshape Compatibility

The total number of elements **must match** the requested shape. For example, an array of size 12 cannot be reshaped into `(5, 3)`, because that shape requires 15 elements.

NumPy can calculate **one** missing dimension when you specify `-1`.

````python
reshaped_automatically = values_12.reshape(3, -1)

print(reshaped_automatically)
print("Inferred shape:", reshaped_automatically.shape)
````

### 4.4 Flattening Arrays with `flatten()`

`.flatten()` converts a multidimensional array into a **new 1D copy**. It is useful when the analysis requires all elements in a single sequence.

````python
original_matrix = np.array([[1, 2],
                            [5, 7],
                            [6, 6]])
flattened_values = original_matrix.flatten()

print("Original:")
print(original_matrix)
print("Flattened:")
print(flattened_values)
print("Flattened shape:", flattened_values.shape)
````

### 4.5 `reshape()` vs. `flatten()`

| Operation | Main purpose | Number of dimensions after operation |
|---|---|---|
| `reshape(new_shape)` | Rearrange elements into a chosen shape | Determined by `new_shape` |
| `flatten()` | Convert all elements into a new 1D array | Always 1D |

## 5. Basic Array Indexing and Slicing

### 5.1 Indexing a One-Dimensional Array (1D)

**Indexing** selects an element based on its position. Python uses **zero-based indexing**, so index `0` is the first element, index `1` is the second, and so on.

Negative indices count backward from the end: `-1` is the last element.

````python
values = np.array([2, 4, 6, 8, 10])

print("First element:", values[0])
print("Fourth element:", values[3])
print("Last element:", values[-1])
````

### 5.2 Indexing a Two-Dimensional Array (2D)

A 2D array requires a **row index** and a **column index** to select one element. The syntax is `array[row, column]`. Both indices start from zero.

````python
marks = np.array([[80, 85, 90],
                  [75, 88, 84],
                  [92, 90, 95]])

print(marks)
print("Row 1, column 2:", marks[1, 2])
print("Row 2, column 0:", marks[2, 0])
````

### 5.3 Selecting Rows and Columns

Use a single index to select a row (`array[row]`) or combine an index with `:` to select a complete column (`array[:, column]`). The colon means **all positions** along that axis.

````python
print("First row:", marks[0])
print("Second row:", marks[1, :])
print("Third column:", marks[:, 2])
print("First column:", marks[:, 0])
````

### 5.4 Slicing a One-Dimensional Array (1D)

**Slicing** retrieves several consecutive elements. The general syntax is `array[start:stop]`.

- `start` is included.
- `stop` is excluded.
- An omitted `start` means the beginning of the array.
- An omitted `stop` means the end of the array.

````python
sequence = np.array([10, 20, 30, 40, 50, 60])

print("Positions 1 to 3:", sequence[1:4])
print("First three:", sequence[:3])
print("From position 3 onward:", sequence[3:])
print("All elements:", sequence[:])
````

### 5.5 Slicing a Two-Dimensional Array (2D)

A 2D slice specifies a range of rows and a range of columns: `array[row_start:row_stop, col_start:col_stop]`. This is useful for extracting a submatrix.

````python
number_grid = np.arange(1, 26).reshape(5, 5)

print("Original matrix:")
print(number_grid)
print("Rows 1–3, columns 2–4:")
print(number_grid[1:4, 2:5])
````

### 5.6 Slicing with Steps

A slice can include a third argument: `array[start:stop:step]`. The `step` specifies the interval between selected positions.

- `::2` selects every second element, starting at the beginning.
- `1::2` selects every second element, starting at index 1.
- `[::-1]` reverses a 1D array.

Steps can also be applied separately to the row and column axes of a 2D array.

````python
print("Every second element:", sequence[::2])
print("Every second from index 1:", sequence[1::2])
print("Reversed:", sequence[::-1])
print("Every second row and column:")
print(number_grid[::2, ::2])
````

### 5.8 Common Indexing and Slicing Mistakes

1. **Off-by-one error:** `a[1:4]` selects indices 1, 2, and 3, not index 4.
2. **Mixing row and column order:** `a[2, 1]` means row index 2 and column index 1.
3. **Confusing element and row selection:** `a[1]` returns a row from a 2D array; `a[1, 1]` returns one element.
4. **Index out of range:** a 5-element array has valid positive indices from 0 through 4.
5. **Confusing 1D and 2D shapes:** a 1D array of size 5 has shape `(5,)`, not `(1, 5)`.

**Conceptual distinction:** Indexing usually selects an individual position; slicing selects a range of positions. More advanced conditional selection and modification of array values will be discussed in Meeting 2.

## 1. Sorting Arrays

### 1.1 Introduction to Array Sorting

**Sorting** adalah proses mengurutkan elemen array, misal dari nilai terkecil ke terbesar. Fungsi `np.sort()` menghasilkan **array baru** yang sudah terurut sehingga array asli tidak berubah.

### 1.2 Sorting a One-Dimensional Array (1D)

Untuk array 1D, `np.sort()` mengurutkan semua elemen secara menaik. Gunakan `[::-1]` pada hasilnya jika ingin urutan menurun.

````python
scores = np.array([88, 72, 95, 80, 76])

print("Original:", scores)
print("Ascending:", np.sort(scores))
print("Descending:", np.sort(scores)[::-1])
print("Original remains:", scores)
````

[::-1] → Membalik urutan array, sehingga menjadi descending.

### 1.3 Sorting a Two-Dimensional Array (2D)

Pada array 2D, `np.sort()` secara default mengurutkan nilai **di setiap baris** (`axis=-1`, sama dengan `axis=1` untuk array 2D). Fungsi ini tidak mengurutkan seluruh baris berdasarkan satu kolom.

````python
measurement_matrix = np.array([[8, 3, 6],
                               [9, 2, 5],
                               [4, 7, 1]])

print("Original matrix:")
print(measurement_matrix)
print("Sorted within each row:")
print(np.sort(measurement_matrix))
````

### 1.4 Sorting by Axis

Untuk array 2D:
- **`axis=0`**: mengurutkan nilai dari atas ke bawah pada **setiap kolom**.
- **`axis=1`**: mengurutkan nilai dari kiri ke kanan pada **setiap baris**.
- **`axis=None`**: mengurutkan seluruh elemen dan menghasilkan array 1D.

````python
print("Sorted by columns (axis=0):")
print(np.sort(measurement_matrix, axis=0))

print("Sorted by rows (axis=1):")
print(np.sort(measurement_matrix, axis=1))

print("All elements sorted:", np.sort(measurement_matrix, axis=None))
````

### 1.5 Sorting Indices with `np.argsort()`

`np.argsort()` menghasilkan **indeks urutan**, bukan nilai yang langsung diurutkan. Indeks tersebut berguna jika nilai dan nama mahasiswa harus tetap berpasangan.

````python
students = np.array(["Ayu", "Budi", "Citra", "Doni"])
final_scores = np.array([81, 94, 76, 88])
order = np.argsort(final_scores)

print("Sorted scores:", final_scores[order])
print("Corresponding students:", students[order])
````

## 2. Filtering Arrays

### 2.1 Understanding Boolean Masks

**Filtering** digunakan untuk memilih data yang memenuhi syarat tertentu. **Boolean mask** adalah array berisi `True` dan `False` yang dihasilkan dari operasi perbandingan.

````python
exam_scores = np.array([65, 78, 84, 92, 71, 88])
mask = exam_scores >= 80

print("Scores:", exam_scores)
print("Mask >= 80:", mask)
````

### 2.2 Boolean Indexing

Gunakan `array[mask]` untuk mengambil elemen pada posisi yang bernilai `True`. Hasilnya hanya berisi elemen yang memenuhi syarat.

````python
print("Scores >= 80:", exam_scores[exam_scores >= 80])
print("Scores < 80:", exam_scores[exam_scores < 80])
````

### 2.3 Boolean Indexing in a Two-Dimensional Array

Untuk matriks, kondisi dapat digunakan untuk memilih **baris** berdasarkan salah satu kolom. Perhatikan bahwa hasil pilihan beberapa baris tetap berupa array 2D.

````python
student_marks = np.array([[75, 82, 80],
                          [90, 93, 88],
                          [68, 74, 79],
                          [85, 87, 91]])

selected_rows = student_marks[student_marks[:, 0] >= 80]
print("Students with first score >= 80:")
print(selected_rows)
print("Selected shape:", selected_rows.shape)
````

````python
student_marks[:, 0]
````

````python
# urutannya adalah slicing dari semua baris (:) kolom pertama (0) yang first score lebih dari 80
# lalu logika boleannya untuk menyaring memilih baris baris sesuai kondisi booelan dengan semua kolom terpilih [kondisi,]
student_marks[student_marks[:, 0] >= 80]
````

### 2.4 Fancy Indexing

**Fancy indexing** memilih elemen menggunakan daftar indeks tertentu. Berbeda dengan Boolean indexing, kita menentukan **posisi** elemen yang ingin diambil.

````python
values = np.array([10, 20, 30, 40, 50])
print("Selected positions:", values[[0, 2, 4]])

print("Selected rows:")
print(student_marks[[0, 3], :])
````

### 2.5 Filtering with `np.where()`

`np.where(kondisi)` menghasilkan **indeks posisi** yang memenuhi kondisi. Sementara itu, `np.where(kondisi, nilai_benar, nilai_salah)` memilih nilai berdasarkan kondisi, bukan hanya memfilter elemen.

````python
rainfall = np.array([0, 12, 8, 25, 4, 18])
print("Indices with rainfall >= 10:", np.where(rainfall >= 10)[0])
print("Values with rainfall >= 10:", rainfall[np.where(rainfall >= 10)])

print("Rain / No rain:", np.where(rainfall > 10, "Rain", "No rain"))
````

````python
np.where(rainfall >= 10)
````

### 2.6 Conditional Selection with Multiple Conditions

Gunakan **`&`** untuk *AND*, **`|`** untuk *OR*, dan **`~`** untuk *NOT*. Setiap kondisi harus diberi tanda kurung. Jangan menggunakan `and` atau `or` untuk membandingkan seluruh array NumPy.

````python
temperatures = np.array([26, 29, 32, 35, 28, 31])

print("28 to 32:", temperatures[(temperatures >= 28) & (temperatures <= 32)])
print("Below 28 or above 32:", temperatures[(temperatures < 28) | (temperatures > 32)])
print("Not above 30:", temperatures[~(temperatures > 30)])
````

## 3. Updating Arrays

### 3.1 Updating Individual Array Elements

Nilai array bisa diubah menggunakan indeks, misalnya `array[1] = 10`. Gunakan `.copy()` jika ingin mengubah array baru tanpa mengubah data aslinya.

````python
original_data = np.array([10, 20, 30, 40])
updated_data = original_data.copy()
updated_data[1] = 25

print("Original:", original_data)
print("Updated:", updated_data)
````

tanpa .copy maka nanti saat mengupdate data nilai original data juga akan berubah

### 3.2 Updating Rows and Columns

Pada array 2D, gunakan `array[baris, kolom]` untuk satu nilai, `array[baris, :]` untuk satu baris, dan `array[:, kolom]` untuk satu kolom. Jumlah nilai pengganti harus sesuai dengan posisi yang dipilih, kecuali jika menggunakan satu nilai skalar.

````python
marks = np.array([[70, 75, 80],
                  [82, 86, 88],
                  [90, 91, 93]])

marks[0, 0] = 78
marks[1, :] = [84, 87, 90]
marks[:, 2] = 95

print("Updated marks:")
print(marks)
````

### 3.3 Conditional Replacement

Nilai yang memenuhi kondisi tertentu dapat diganti sekaligus dengan **Boolean indexing** atau `np.where()`. Untuk menghindari perubahan data asli, salin array terlebih dahulu.

````python
measurements = np.array([12, -2, 18, -5, 9])
cleaned = measurements.copy()
cleaned[cleaned < 0] = 0

print("Original:", measurements)
print("Updated with mask:", cleaned)
print("Updated with np.where:", np.where(measurements < 0, 0, measurements))
````

np.where(measurements < 0, 0, measurements) artinya jika <0 diberi nilai 0 jika tidak dikembalikan ke nilai aslinya

### 3.4 Adding Rows with `np.concatenate()`

`np.concatenate((a, b), axis=0)` menggabungkan **baris**. Untuk matriks, jumlah **kolom** kedua array harus sama. Fungsi ini menghasilkan array baru.

````python
sales = np.array([[12, 15],
                  [10, 18]])
new_row = np.array([[14, 16]])

sales_with_row = np.concatenate((sales, new_row), axis=0)
print("Added row:")
print(sales_with_row)
print("New shape:", sales_with_row.shape)
````

### 3.5 Adding Columns with `np.concatenate()`

Gunakan `axis=1` untuk menggabungkan **kolom**. Jumlah **baris** harus sama. Kolom baru perlu berbentuk 2D, misalnya `(3, 1)`, bukan `(3,)`.

````python
new_column = np.array([[20], [21], [22]])
sales_with_column = np.concatenate((sales_with_row, new_column), axis=1)

print("Added column:")
print(sales_with_column)
print("New shape:", sales_with_column.shape)
````

### 3.6 Deleting Rows and Columns with `np.delete()`

`np.delete(array, indeks, axis=0)` menghapus **baris**, sedangkan `axis=1` menghapus **kolom**. Hasilnya adalah array baru; array asli tetap tersedia.

````python
print("Delete second row:")
print(np.delete(sales_with_column, 1, axis=0))
print("Delete third column:")
print(np.delete(sales_with_column, 2, axis=1))
````

````python
sales_with_column
````

### 3.7 Shape and Dimension Compatibility

Saat menggabungkan array, dimensi (`ndim`) harus sama. Untuk `axis=0`, jumlah kolom harus sama; untuk `axis=1`, jumlah baris harus sama. `reshape()` membantu mengubah vektor 1D menjadi baris atau kolom 2D.

````python
matrix = np.arange(1, 7).reshape(2, 3)
row_values = np.array([7, 8, 9]).reshape(1, 3)
column_values = np.array([10, 11]).reshape(2, 1)

print("Matrix shape:", matrix.shape)
print("New row shape:", row_values.shape)
print("New column shape:", column_values.shape)

print("With row:")
print(np.concatenate((matrix, row_values), axis=0))

print("With column:")
print(np.concatenate((matrix, column_values), axis=1))
````

**Common mistakes:** Menambahkan kolom berbentuk `(2,)` ke matriks `(2, 3)` dengan `axis=1` tidak dapat dilakukan langsung. Ubah dahulu menjadi `(2, 1)`. Jangan menggunakan `np.delete()` tanpa `axis` jika ingin mempertahankan bentuk matriks karena hasil default-nya diratakan menjadi 1D.

## 4. Array Aggregation

### 4.1 Sum, Minimum, Maximum, and Mean

**Aggregation** merangkum banyak nilai menjadi satu atau beberapa statistik. Fungsi dasar yang sering dipakai adalah `np.sum()`, `np.min()`, `np.max()`, dan `np.mean()`.

````python
daily_visitors = np.array([120, 135, 128, 142, 150])

print("Total:", np.sum(daily_visitors))
print("Minimum:", np.min(daily_visitors))
print("Maximum:", np.max(daily_visitors))
print("Mean:", np.mean(daily_visitors))
````

### 4.2 Aggregation by Axis

Pada matriks, **`axis=0`** meringkas nilai dari berbagai baris untuk **setiap kolom**; **`axis=1`** meringkas nilai dari berbagai kolom untuk **setiap baris**.

**axis=0** totalan/rataan/... semua baris hasilnya per masing2 kolom ... per variabel

**axis=1** totalan/rataan/... semua kolom hasilnya per masing2 baris ... per observasi

````python
test_scores = np.array([[80, 85, 90],
                        [70, 75, 80],
                        [90, 92, 94]])

print("Total per test (axis=0):", np.sum(test_scores, axis=0))
print("Mean per student (axis=1):", np.mean(test_scores, axis=1))
print("Maximum per test (axis=0):", np.max(test_scores, axis=0))
````

### 4.3 Cumulative Sum

`np.cumsum()` menghitung jumlah **kumulatif** secara bertahap. Nilai di setiap posisi adalah jumlah seluruh nilai sampai posisi tersebut. `axis` juga dapat digunakan untuk data 2D.

````python
weekly_sales = np.array([10, 15, 12, 18])

print("Weekly sales:", weekly_sales)
print("Cumulative sales:", np.cumsum(weekly_sales))

print("Cumulative scores across columns:")
print(np.cumsum(test_scores, axis=1))
````

### 4.4 The `keepdims` Argument

Secara default, agregasi menghilangkan sumbu yang diringkas. `keepdims=True` mempertahankan sumbu tersebut dengan ukuran **1**, sehingga hasil lebih mudah dipakai untuk operasi lanjutan yang membutuhkan bentuk kompatibel.

````python
row_means = np.mean(test_scores, axis=1)
row_means_2d = np.mean(test_scores, axis=1, keepdims=True)

print("Without keepdims:", row_means, "shape:", row_means.shape)
print("With keepdims:")
print(row_means_2d)
print("Shape:", row_means_2d.shape)
````

### 4.5 Applying Aggregation to Statistical Data

Misalkan setiap baris adalah satu hari dan setiap kolom adalah satu lokasi pengamatan curah hujan. Agregasi dapat menghitung total per lokasi dan rata-rata per hari.

````python
rainfall_matrix = np.array([[12, 8, 5],
                            [10, 15, 7],
                            [20, 11, 9],
                            [8, 6, 4]])

print("Total per location:", np.sum(rainfall_matrix, axis=0))
print("Mean per day:", np.mean(rainfall_matrix, axis=1))
print("Overall maximum:", np.max(rainfall_matrix))
````

## 5. Vectorized Operations

### 5.1 Vectorization vs. Python Loops

**Vectorization** memungkinkan perhitungan langsung pada seluruh array tanpa menulis perulangan untuk setiap elemen. Kode umumnya lebih singkat dan sering lebih efisien untuk data numerik.

````python
values = np.array([2, 4, 6, 8])

# Python loop
loop_result = []
for value in values:
    loop_result.append(value * 2)

# NumPy vectorization
vector_result = values * 2

print("Loop:", loop_result)
print("Vectorized:", vector_result)
````

### 5.2 Scalar Arithmetic Operations

Satu bilangan (**skalar**) dapat dijumlahkan, dikurangkan, dikalikan, atau dibagikan ke setiap elemen array. Operasi ini bekerja untuk array 1D maupun 2D.

````python
data = np.array([10, 20, 30])

print("Add 5:", data + 5)
print("Subtract 2:", data - 2)
print("Multiply by 3:", data * 3)
print("Divide by 2:", data / 2)
print("Squared:", data ** 2)
````

### 5.3 Element-wise Array Mathematics

**Element-wise** berarti operasi dilakukan pada elemen yang **posisinya bersesuaian**. Contohnya, `a + b` menjumlahkan elemen pertama dengan pertama, kedua dengan kedua, dan seterusnya. Kedua array pada contoh berikut memiliki bentuk sama.

````python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Powers:", a ** 2)
````

### 5.4 Element-wise Comparison Operations

Operator `>`, `<`, `>=`, `<=`, `==`, dan `!=` membandingkan setiap elemen dan menghasilkan array Boolean. Hasil ini dapat dipakai kembali untuk filtering atau pembaruan nilai.

````python
actual = np.array([75, 90, 82, 60])
target = np.array([80, 85, 80, 70])

print("Actual >= target:", actual >= target)
print("Actual == target:", actual == target)
print("Passed target:", actual[actual >= target])
````

### 5.5 Shape Compatibility and Broadcasting

Operasi element-wise memerlukan bentuk yang **kompatibel**. **Broadcasting** memungkinkan array dengan bentuk berbeda dipasangkan apabila ukurannya sama atau salah satunya 1 saat dibandingkan dari dimensi paling kanan. Contohnya, vektor tiga nilai dapat ditambahkan pada setiap baris matriks dengan tiga kolom.

````python
matrix = np.array([[10, 20, 30],
                   [40, 50, 60]])
adjustment = np.array([1, 2, 3])

print("Matrix shape:", matrix.shape)
print("Adjustment shape:", adjustment.shape)
print("Element-wise result:")
print(matrix + adjustment)
````

**Important distinction:** Operator `*` pada dua array adalah perkalian **element-wise**, bukan perkalian matriks. Untuk perkalian matriks gunakan `@` atau `np.matmul()`. Operasi dengan bentuk yang tidak kompatibel akan menghasilkan `ValueError`.

### 5.6 Applied Example: Standardizing Daily Measurements

Contoh sederhana berikut menghitung selisih setiap pengamatan dari rata-ratanya. Karena menggunakan operasi vektorisasi, kita tidak perlu menulis perulangan.

````python
observations = np.array([12.0, 15.0, 13.0, 20.0])
mean_value = np.mean(observations)
deviations = observations - mean_value

print("Mean:", mean_value)
print("Deviations:", deviations)
print("Sum of deviations:", np.sum(deviations))
````

## 1. Broadcasting in NumPy

### 1.1 Introduction to Broadcasting

**Broadcasting** memungkinkan operasi antar-array dengan bentuk (*shape*) berbeda tanpa mengulang data secara manual. NumPy menyesuaikan dimensi yang kompatibel saat melakukan operasi **element-wise**. Contoh paling sederhana adalah menambahkan satu bilangan ke seluruh elemen array.

````python
measurements = np.array([10, 20, 30, 40])
correction = 5

print("Original:", measurements)
print("Correction:", correction)
print("After broadcasting:", measurements + correction)
print("Original remains:", measurements)
````

### 1.2 Broadcasting Rules

NumPy memeriksa ukuran setiap dimensi array dari kanan ke kiri. Dua array dapat dioperasikan jika ukuran dimensinya **sama** atau **salah satunya bernilai** 1. Jika suatu dimensi tidak ada, NumPy menganggap ukurannya 1. Jika ukuran dimensi tidak memenuhi aturan tersebut, operasi akan gagal dan muncul pesan kesalahan `ValueError`.

````python
matrix = np.array([[10, 20, 30],
                   [40, 50, 60]])
row_adjustment = np.array([1, 2, 3])

print("Matrix shape:", matrix.shape)
print("Adjustment shape:", row_adjustment.shape)
print("Broadcast result shape:", (matrix + row_adjustment).shape)
print(matrix + row_adjustment)

# Contoh yang tidak kompatibel: (2, 3) dan (2,)
incompatible = np.array([1, 2])
try:
    print(matrix + incompatible)
except ValueError as error:
    print("Incompatible shapes:", error)
````

### 1.3 Shape Compatibility

Bentuk `(3, 1)` memiliki 3 baris dan 1 kolom, sedangkan `(1, 4)` memiliki 1 baris dan 4 kolom. Melalui broadcasting, NumPy secara otomatis menyesuaikan dimensi berukuran 1, sehingga menghasilkan bentuk `(3, 4)`. Fungsi `np.broadcast_shapes()` digunakan untuk memeriksa bentuk hasil sebelum operasi dilakukan.

````python
column_vector = np.array([[10],
                          [20],
                          [30]])
row_vector = np.array([[1, 2, 3, 4]])

combined = column_vector + row_vector

print("Column shape:", column_vector.shape)
print("Row shape:", row_vector.shape)
print("Compatible result shape:", np.broadcast_shapes(column_vector.shape, row_vector.shape))
print("Combined array:")
print(combined)
````

### 1.4 Broadcasting Across Rows and Columns

Pada array 2D, kita dapat menambahkan nilai yang berbeda untuk setiap **kolom** menggunakan vektor berbentuk `(n_columns,)`. Jika ingin menambahkan nilai yang berbeda untuk setiap **baris**, gunakan `reshape(-1, 1)` untuk mengubah vektor menjadi bentuk `(n_rows, 1)`.

````python
scores = np.array([[70, 80, 90],
                   [75, 85, 95],
                   [60, 70, 80]])

column_bonus = np.array([2, 3, 5])
row_bonus = np.array([1, 4, 2]).reshape(-1, 1)

print("Bonus per column (applies to each row):")
print(scores + column_bonus)
print("Bonus per row (applies to each column):")
print(scores + row_bonus)
print("Row bonus shape:", row_bonus.shape)
````

**Applied example: Centering by Column Mean**

Untuk mengurangi setiap nilai data dengan **rata-rata kolomnya**, gunakan `np.mean(..., axis=0, keepdims=True)`. Hasil rata-rata berbentuk `(1, n_columns)`, sehingga melalui **broadcasting**, setiap nilai pada kolom akan dikurangi dengan rata-rata kolom tersebut.

````python
variable_means = np.mean(scores, axis=0, keepdims=True)
centered_scores = scores - variable_means

print("Column means:", variable_means)
print("Column means shape:", variable_means.shape)
print("Centered scores:")
print(centered_scores)
print("Means after centering:", np.mean(centered_scores, axis=0))
````

**Common mistakes:** Array berbentuk `(3,)` tidak otomatis dianggap sebagai kolom `(3, 1)`. NumPy membandingkan ukuran dimensi dari **kanan ke kiri**. Jika ingin menggunakan nilai yang berbeda untuk setiap **baris**, ubah bentuk array menjadi `(3, 1)` menggunakan `reshape(-1, 1)`.

## 3. Image Arrays and Transformations

### 3.1 Understanding RGB and RGBA Arrays

Gambar berwarna dapat direpresentasikan sebagai array 3D dengan bentuk **`(height, width, channels)`**. **RGB** memiliki 3 kanal (Red, Green, Blue); **RGBA** memiliki kanal keempat, yaitu Alpha (transparansi). Pada gambar 8-bit, nilai tiap kanal biasanya berada pada rentang **0 sampai 255**.

````python
# Gambar RGB sederhana: 3 baris x 3 kolom x 3 kanal warna
rgb_image = np.array([
    [[255, 0, 0],   [0, 255, 0],   [0, 0, 255]],
    [[255, 255, 0], [0, 255, 255], [255, 0, 255]],
    [[255, 255, 255], [128, 128, 128], [0, 0, 0]]
], dtype=np.uint8)

alpha_channel = np.full((*rgb_image.shape[:2], 1), 255, dtype=np.uint8)
rgba_image = np.concatenate((rgb_image, alpha_channel), axis=2)

print("RGB shape:", rgb_image.shape)
print("RGBA shape:", rgba_image.shape)
print("Top-left RGB pixel:", rgb_image[0, 0])
print("Top-left RGBA pixel:", rgba_image[0, 0])
````

### 3.2 Selecting Image Color Channels

Gunakan `image[:, :, 0]` untuk kanal **Red**, indeks `1` untuk **Green**, dan indeks `2` untuk **Blue**. Setelah memilih satu kanal, hasilnya menjadi array **2D** `(height, width)`. Pemilihan `image[:, :, 0:1]` mempertahankan dimensi kanal.

````python
red_channel = rgb_image[:, :, 0]
green_channel = rgb_image[:, :, 1]
blue_channel = rgb_image[:, :, 2]

print("Red channel:")
print(red_channel)
print("Green channel:")
print(green_channel)
print("Blue channel:")
print(blue_channel)
print("Red shape:", red_channel.shape)
print("Red shape with 0:1:", rgb_image[:, :, 0:1].shape)
````

### 3.3 Manipulating Pixel Values

Nilai piksel dapat diubah dengan operasi NumPy, misalnya meningkatkan kecerahan atau mengurangi salah satu kanal. Karena `uint8` hanya menampung nilai 0–255, ubah dahulu ke tipe bilangan yang lebih besar sebelum penjumlahan, kemudian batasi hasil dengan **`np.clip()`** agar tidak terjadi *overflow*.

````python
# Gunakan tipe lebih besar untuk menghindari overflow uint8
brightened_image = np.clip(rgb_image.astype(np.int16) + 40, 0, 255).astype(np.uint8)

less_red_image = rgb_image.copy()
less_red_image[:, :, 0] = 0

print("Original pixel [1, 1]:", rgb_image[1, 1])
print("Brighter pixel [1, 1]:", brightened_image[1, 1])
print("Original pixel [0, 0]:", rgb_image[0, 0])
print("Pixel without red [0, 0]:", less_red_image[0, 0])
print("Original image remains:", rgb_image[0, 0])
````

### 3.4 Flipping Arrays (`np.flip`)

`np.flip(image, axis=0)` membalik posisi **atas–bawah** (vertikal), sedangkan `axis=1` membalik posisi **kiri–kanan** (horizontal). Pada gambar RGB, sumbu `2` adalah kanal warna, sehingga membalik `axis=2` justru mengubah urutan kanal, bukan orientasi gambar.

````python
vertical_flip = np.flip(rgb_image, axis=0)
horizontal_flip = np.flip(rgb_image, axis=1)

print("Original red channel:")
print(rgb_image[:, :, 0])
print("Vertical flip (axis=0):")
print(vertical_flip[:, :, 0])
print("Horizontal flip (axis=1):")
print(horizontal_flip[:, :, 0])
print("Flipped shape:", horizontal_flip.shape)
````

## 4. Transposing Arrays

### 4.1 Array Transposition

**Transposition** menukar susunan sumbu array. Untuk matriks 2D, `.T` atau `np.transpose()` mengubah baris menjadi kolom dan sebaliknya: `(rows, columns)` menjadi `(columns, rows)`. Pada array 1D, `.T` tidak mengubah bentuknya.

````python
data_matrix = np.array([[1, 2, 3],
                        [4, 5, 6]])

print("Original:")
print(data_matrix)
print("Shape:", data_matrix.shape)
print("Transposed:")
print(data_matrix.T)
print("Transposed shape:", np.transpose(data_matrix).shape)
print("1D transpose shape:", np.array([1, 2, 3]).T.shape)
````

### 4.2 Transposing by Axis

Untuk array 3D atau lebih, `np.transpose(array, axes=(...))` menentukan **urutan baru semua sumbu**. Misalnya, `(0, 1, 2)` menjadi `(1, 0, 2)` berarti menukar sumbu pertama dan kedua. `np.swapaxes()` dapat dipakai jika hanya dua sumbu yang ditukar.

````python
data_3d = np.arange(24).reshape(2, 3, 4)
reordered = np.transpose(data_3d, axes=(1, 0, 2))

print("=== ORIGINAL ARRAY ===")
print("Shape:", data_3d.shape)
print(data_3d)

print("\nLayer 0:")
print(data_3d[0])

print("\nLayer 1:")
print(data_3d[1])

print("\n=== TRANSPOSED ARRAY ===")
print("Shape:", reordered.shape)
print(reordered)

print("\nLayer 0:")
print(reordered[0])

print("\nLayer 1:")
print(reordered[1])

print("\nLayer 2:")
print(reordered[2])

print("\n=== ELEMENT COMPARISON ===")
print("Original [1, 2, 3]:", data_3d[1, 2, 3])
print("Transposed [2, 1, 3]:", reordered[2, 1, 3])

print("\n=== SWAPAXES ===")
print(np.swapaxes(data_3d, 0, 1))
````

## 5. Stacking and Splitting Arrays

### 5.1 Splitting Arrays (`np.split`)

**Splitting** digunakan untuk membagi satu array menjadi beberapa bagian. Fungsi `np.split()` membagi array dengan jumlah elemen yang **sama banyak** pada setiap bagian. Jika jumlah elemen tidak dapat dibagi rata, gunakan `np.array_split()` yang memungkinkan hasil pembagian memiliki jumlah elemen berbeda.

````python
monthly_values = np.array([10, 12, 15, 18, 20, 22])
first_half, second_half = np.split(monthly_values, 2)

print("First half:", first_half)
print("Second half:", second_half)

measurement_table = np.arange(1, 13).reshape(3, 4)
left_part, right_part = np.split(measurement_table, 2, axis=1)

print("Original table:")
print(measurement_table)
print("Left half shape:", left_part.shape)
print(left_part)
print("Right half shape:", right_part.shape)
print(right_part)
print("Unequal split example:", [part.tolist() for part in np.array_split(np.arange(7), 3)])
````

Kode `np.arange(7)` membuat array berisi angka `[0, 1, 2, 3, 4, 5, 6]`. Selanjutnya, `np.array_split(..., 3)` membagi array tersebut menjadi **3 bagian** dengan jumlah elemen masing-masing **3, 2, dan 2**. Perulangan `for part in ...` mengambil setiap bagian, kemudian `part.tolist()` mengubahnya menjadi **list Python**. Terakhir, `print()` menampilkan hasil pembagian berupa `[[0, 1, 2], [3, 4], [5, 6]]`.

### 5.2 Stacking Arrays (`np.stack`)

**Stacking** menyatukan array yang memiliki bentuk **sama** dengan membuat **sumbu baru**. Sebagai contoh, dua vektor `(3,)` dapat menjadi `(2, 3)` jika `axis=0` atau `(3, 2)` jika `axis=1`. Berbeda dari `np.concatenate()`, `np.stack()` menambah jumlah dimensi, bisa dilihat dari dimensi/ notasi kurung sebelum di stack atau di concatenate

````python
day_one = np.array([12, 15, 18])
day_two = np.array([10, 17, 20])

stacked_rows = np.stack((day_one, day_two), axis=0)
stacked_columns = np.stack((day_one, day_two), axis=1)

print("Original vector shape:", day_one.shape)
print("Stacked by rows:")
print(stacked_rows)
print("Shape:", stacked_rows.shape)
print("Stacked by columns:")
print(stacked_columns)
print("Shape:", stacked_columns.shape)
````

### 5.3 Reshaping Split Arrays

Setelah array dibagi menggunakan `np.split()`, setiap bagian dapat diubah bentuknya menggunakan `reshape()` tanpa mengubah **jumlah elemen**. Misalnya, array 2D berbentuk `(3, 2)` yang memiliki 6 elemen dapat diubah menjadi array 1D berbentuk `(6,)`. Selanjutnya, dua array 1D tersebut dapat digabungkan menggunakan `np.stack()` menjadi array 2D berbentuk `(2, 6)`.

````python
first_block, second_block = np.split(measurement_table, 2, axis=1)
first_vector = first_block.reshape(-1)
second_vector = second_block.reshape(-1)
reshaped_and_stacked = np.stack((first_vector, second_vector), axis=0)

print("Shape of each split block:", first_block.shape)
print("Shape after reshape:", first_vector.shape)
print("Reshaped and stacked:")
print(reshaped_and_stacked)
print("New shape:", reshaped_and_stacked.shape)
````

## Answer Key — Jawaban Independent Student Exercises

### Exercise 1 — Broadcasting in NumPy

````python
scores = np.array([[70, 80, 90],
                   [75, 85, 95],
                   [60, 70, 80]])

# 1. Bonus per kolom
col_bonus = np.array([5, 3, 2])
result_col = scores + col_bonus

# 2. Bonus per baris (reshape jadi kolom (3, 1))
row_bonus = np.array([1, 4, 2]).reshape(-1, 1)
result_row = scores + row_bonus

# 3. Centering dengan rata-rata kolom
col_mean = np.mean(scores, axis=0, keepdims=True)
centered = scores - col_mean

print("Bonus per kolom:\n", result_col)
print("Bonus per baris:\n", result_row)
print("Rata-rata kolom:", col_mean)
print("Hasil centering:\n", centered)

# 4. Shape
print("\nShape col_bonus :", col_bonus.shape)
print("Shape row_bonus :", row_bonus.shape)
print("Shape col_mean  :", col_mean.shape)
print("Shape hasil     :", result_col.shape, result_row.shape, centered.shape)
````

### Exercise 2 — Saving and Loading Arrays

````python
temperatures = np.array([[29, 31, 30], [28, 32, 33], [27, 30, 31]])

# 1. Simpan
np.save("week11_temperatures.npy", temperatures)

# 2. Muat kembali
loaded_temperatures = np.load("week11_temperatures.npy", allow_pickle=False)

# 3. Periksa struktur
print("Shape :", loaded_temperatures.shape)
print("ndim  :", loaded_temperatures.ndim)
print("dtype :", loaded_temperatures.dtype)

# 4. Bandingkan
print("Sama dengan aslinya:", np.array_equal(temperatures, loaded_temperatures))
````

### Exercise 3 — Image Arrays and Transformations

````python
image = np.array([[[255, 0, 0], [0, 255, 0]],
                  [[0, 0, 255], [255, 255, 255]]], dtype=np.uint8)

# 1. Bentuk dan kanal
print("Shape:", image.shape)
print("Red:\n", image[:, :, 0])
print("Green:\n", image[:, :, 1])
print("Blue:\n", image[:, :, 2])

# 2. Hapus kanal Blue pada salinan
no_blue = image.copy()
no_blue[:, :, 2] = 0
print("\nTanpa Blue:\n", no_blue)

# 3. Tambah kecerahan 30 (hindari overflow uint8)
brighter = np.clip(image.astype(np.int16) + 30, 0, 255).astype(np.uint8)
print("\nLebih cerah:\n", brighter)

# 4. Flip dan cek piksel [0, 0]
h_flip = np.flip(image, axis=1)   # kiri-kanan
v_flip = np.flip(image, axis=0)   # atas-bawah
print("\nPiksel [0,0] asli      :", image[0, 0])
print("Piksel [0,0] horizontal:", h_flip[0, 0])
print("Piksel [0,0] vertikal  :", v_flip[0, 0])
````

### Exercise 4 — Transposing Arrays

````python
arr = np.arange(24).reshape(2, 3, 4)

# 1. Shape awal
print("Shape awal:", arr.shape)

# 2. transpose urutan sumbu (1, 0, 2)
transposed = np.transpose(arr, axes=(1, 0, 2))
print("transpose  :", transposed.shape)

# 3. swapaxes menghasilkan bentuk yang sama
swapped = np.swapaxes(arr, 0, 1)
print("swapaxes   :", swapped.shape, "| isi sama:", np.array_equal(transposed, swapped))

# 4. HWC -> CHW -> HWC
image = np.zeros((4, 5, 3))
chw = np.transpose(image, (2, 0, 1))
hwc = np.transpose(chw, (1, 2, 0))
print("HWC :", image.shape)
print("CHW :", chw.shape)
print("HWC kembali:", hwc.shape)
````

### Exercise 5 — Stacking and Splitting Arrays

````python
data = np.arange(1, 13).reshape(3, 4)

# 1. Split menurut kolom
left, right = np.split(data, 2, axis=1)

# 2. Shape lalu jadikan 1D
print("Shape left :", left.shape)
print("Shape right:", right.shape)
left_vec = left.reshape(-1)
right_vec = right.reshape(-1)
print("Shape vektor:", left_vec.shape, right_vec.shape)

# 3. Stack
stacked = np.stack((left_vec, right_vec), axis=0)
print("Stacked:\n", stacked)
print("Shape stacked:", stacked.shape)

# 4. array_split untuk pembagian tidak rata
parts = np.array_split(np.arange(10), 3)
print("\nBagian:", [p.tolist() for p in parts])
print("Ukuran:", [len(p) for p in parts])
````

### Exercise 6 — Integrated Statistical Practice

````python
rainfall = np.array([[12, 8, 5],
                     [10, 15, 7],
                     [20, 11, 9],
                     [8, 6, 4]])

# 1. Rata-rata tiap lokasi + centering
loc_mean = np.mean(rainfall, axis=0, keepdims=True)
centered_rainfall = rainfall - loc_mean
print("Rata-rata per lokasi:", loc_mean)
print("Centered:\n", centered_rainfall)

# 2. Simpan dan muat kembali
np.save("week11_centered_rainfall.npy", centered_rainfall)
loaded_centered = np.load("week11_centered_rainfall.npy", allow_pickle=False)

# 3. Transpose (hari, lokasi) -> (lokasi, hari)
transposed_rain = loaded_centered.T
print("\nShape setelah .T:", transposed_rain.shape)

# 4. Split menurut kolom
part_a, part_b = np.split(transposed_rain, 2, axis=1)
print("Shape bagian:", part_a.shape, part_b.shape)

# 5. Gabungkan kembali dan buktikan
rejoined = np.concatenate((part_a, part_b), axis=1)
print("Shape gabungan:", rejoined.shape)
print("Sama dengan sebelum dibagi:", np.array_equal(rejoined, transposed_rain))
````

## 7. Meeting Summary and Self-Assessment

### Essential Concepts

- **Broadcasting** memungkinkan operasi antar-array dengan bentuk berbeda jika dimensi dari kanan cocok atau salah satunya 1.
- **`reshape(-1, 1)`** membentuk vektor kolom; **`keepdims=True`** mempertahankan dimensi untuk operasi broadcasting.
- **`np.save()`** menyimpan satu array ke file **`.npy`**; **`np.load()`** membacanya kembali beserta `shape` dan `dtype`.
- **RGB** mempunyai tiga kanal dan **RGBA** empat kanal; bentuk gambar lazimnya `(height, width, channels)`.
- **`image[:, :, k]`** mengambil satu kanal warna; **`np.clip()`** membatasi nilai piksel dan mencegah hasil di luar rentang yang diinginkan.
- **`np.flip(axis=0)`** membalik atas–bawah; **`np.flip(axis=1)`** membalik kiri–kanan.
- **`.T`** mentransposisikan array; **`np.transpose(..., axes=...)`** mengatur ulang sumbu secara eksplisit, termasuk HWC → CHW.
- **`np.split()`** membagi array pada sumbu tertentu; **`np.array_split()`** dapat membagi menjadi ukuran tidak sama.
- **`np.stack()`** menambah sumbu baru, sedangkan **`np.concatenate()`** menggabungkan array pada sumbu yang sudah ada.
