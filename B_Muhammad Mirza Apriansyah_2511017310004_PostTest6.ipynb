## 09 - Introduction to NumPy & Basic Array Selection

**Tujuan:** membuat dan memeriksa array, mengubah tipe serta bentuk array, kemudian memilih elemen atau subset data dengan indexing dan slicing.

### Sub Konteks 01: Pengenalan NumPy, `np.array()`, dan `type()`
NumPy (*Numerical Python*) menyediakan `ndarray` untuk komputasi numerik. Operator `+` pada **list** menggabungkan list, sedangkan pada array NumPy menjumlahkan elemen yang bersesuaian.

```python
# Mengimpor NumPy dan membandingkan list dengan array
import numpy as np

nilai_list = [70, 80, 90]
nilai_array = np.array(nilai_list)

print("Versi NumPy:", np.__version__)
print("Tipe list:", type(nilai_list))
print("Tipe array:", type(nilai_array))
print("List + list:", nilai_list + nilai_list)
print("Array + array:", nilai_array + nilai_array)
```

### Sub Konteks 02: Membuat Array 1D dan 2D
Array 1D merepresentasikan satu variabel. Array 2D berbentuk tabel yang terdiri atas baris dan kolom. Dalam data nilai, satu baris dapat mewakili satu mahasiswa dan satu kolom mewakili satu jenis penilaian.

```python
# Array 1D: nilai ujian lima mahasiswa
nilai_1d = np.array([75, 80, 92, 86, 88])

# Array 2D: 3 mahasiswa x 3 komponen nilai
nilai_2d = np.array([[80, 85, 90],
                     [75, 88, 84],
                     [92, 90, 95]])

print("1D:", nilai_1d, "| ndim:", nilai_1d.ndim, "| shape:", nilai_1d.shape)
print("2D:\n", nilai_2d)
print("ndim:", nilai_2d.ndim, "| shape:", nilai_2d.shape)
```

### Sub Konteks 03: Array 3D dan 4D
Array 3D dapat memuat beberapa matriks, misalnya data menurut minggu, mahasiswa, dan jenis pengukuran. Array 4D menambah satu tingkat pengelompokan. Banyak dimensi **tidak sama** dengan jumlah elemen.

```python
# Bentuk 3D: 2 minggu x 2 mahasiswa x 2 pengukuran
minggu_1 = [[10, 12], [14, 16]]
minggu_2 = [[11, 13], [15, 17]]
data_3d = np.array([minggu_1, minggu_2])

# Bentuk 4D: 2 kelompok x 2 minggu x 2 mahasiswa x 2 pengukuran
data_4d = np.array([data_3d, data_3d + 5])

print("3D shape:", data_3d.shape, "| ndim:", data_3d.ndim)
print("4D shape:", data_4d.shape, "| ndim:", data_4d.ndim)
print("Pengukuran minggu 2, mahasiswa 1, kolom 2:", data_3d[1, 0, 1])
```

### Sub Konteks 04: Membuat Array dengan `np.zeros()`, `np.random.random()`, dan `np.arange()`
`np.zeros()` mengisi array dengan nol, `np.random.random()` menghasilkan bilangan acak pada interval `[0, 1)`, dan `np.arange(start, stop, step)` membuat deret dengan batas `stop` **tidak ikut**. `seed` menjaga keterulangan hasil simulasi.

```python
# Membuat array dengan tiga konstruktor
matriks_nol = np.zeros((2, 3))
np.random.seed(42)
matriks_acak = np.random.random((2, 3))
deret = np.arange(2, 13, 2)

print("Array nol:\n", matriks_nol)
print("Array acak:\n", np.round(matriks_acak, 3))
print("Deret bilangan:", deret)                 # 2, 4, 6, 8, 10, 12
print("Bentuk array acak:", matriks_acak.shape)
print("Bentuk deret:", deret.shape)
```

### Sub Konteks 05: Atribut Array, `axis`, dan Tipe Data
`.ndim` menunjukkan jumlah dimensi, `.shape` menunjukkan ukuran setiap dimensi, `.size` menunjukkan jumlah elemen, dan `.dtype` menunjukkan tipe data. Dalam perhitungan rata-rata, `axis=0` menghasilkan **rata-rata tiap kolom**, sedangkan `axis=1` menghasilkan **rata-rata tiap baris**. Gunakan `dtype=` atau `.astype()` untuk mengatur tipe data. Konversi `float` ke `int` membuang pecahan, bukan membulatkan.

```python
# Atribut array dan perhitungan menurut axis
pengamatan = np.array([[12, 15, 18],
                       [14, 16, 20]])
print("ndim:", pengamatan.ndim, "shape:", pengamatan.shape)
print("size:", pengamatan.size, "dtype:", pengamatan.dtype)
print("Rata-rata tiap kolom:", pengamatan.mean(axis=0))
print("Rata-rata tiap baris:", pengamatan.mean(axis=1))

# dtype eksplisit, astype, dan konversi otomatis (type coercion)
nilai_teks = np.array(["70", "85", "90"])
nilai_angka = nilai_teks.astype(int)
pecahan = np.array([2.9, 3.4], dtype=np.float32)
campuran = np.array([True, False, 5])

print("Teks menjadi integer:", nilai_angka, nilai_angka.dtype)
print("float32 -> int:", pecahan.astype(int))
print("Boolean + integer:", campuran, campuran.dtype)
print("Integer + pecahan:", np.array([10, 20.5]).dtype)
print("Integer + teks:", np.array([1, "dua"]).dtype)
```

### Sub Konteks 06: Mengubah Bentuk Array dengan `.reshape()`
`reshape()` mengubah susunan elemen **tanpa mengubah jumlahnya**. Array 1D berukuran enam memiliki `shape=(6,)`, sedangkan vektor baris dan kolom 2D masing-masing memiliki bentuk `(1, 6)` dan `(6, 1)`. Satu nilai `-1` dapat meminta NumPy menghitung ukuran dimensi secara otomatis.

```python
# Perbandingan bentuk 1D, 2D, 3D dan dimensi -1
angka = np.arange(1, 13)
matriks_3x4 = angka.reshape(3, 4)
blok_3d = angka.reshape(2, 2, 3)

print("1D:", angka.shape)
print("Vektor baris:", angka.reshape(1, 12).shape)
print("Vektor kolom:", angka.reshape(12, 1).shape)
print("Matriks 3x4:\n", matriks_3x4)
print("Array 3D:", blok_3d.shape)
print("Inferensi reshape(3, -1):", angka.reshape(3, -1).shape)

# Ukuran baru harus memuat tepat 12 elemen.
try:
    angka.reshape(5, 3)
except ValueError as error:
    print("Contoh reshape tidak cocok:", type(error).__name__)
```

### Sub Konteks 07: Mengubah Array Menjadi 1D dengan `.flatten()`
`flatten()` selalu mengembalikan salinan array 1D. Sebaliknya, `reshape()` menghasilkan susunan baru sesuai bentuk yang diminta, dan **dapat** berbagi memori dengan array asal.

```python
# Perbandingan reshape() dan flatten()
matriks = np.array([[1, 2, 3],
                    [4, 5, 6]])
data_datar = matriks.flatten()
bentuk_baru = matriks.reshape(3, 2)

print("Matriks awal:\n", matriks)
print("Hasil flatten:", data_datar, "| shape:", data_datar.shape)
print("Hasil reshape:\n", bentuk_baru)

data_datar[0] = 99
print("Elemen awal matriks tetap:", matriks[0, 0])  # 1, karena flatten membuat salinan
```

### Sub Konteks 08: Indexing Array 1D dan 2D
Indeks NumPy dimulai dari `0`. `a[-1]` menunjuk elemen terakhir. Pada array 2D, penulisan `a[baris, kolom]` memilih satu nilai berdasarkan dua indeks.

```python
# Mengambil nilai berdasarkan posisi
nilai = np.array([72, 80, 88, 90, 76])
matriks_nilai = np.array([[80, 85, 90],
                          [75, 88, 84],
                          [92, 90, 95]])

print("Elemen pertama:", nilai[0])
print("Elemen terakhir:", nilai[-1])
print("Baris kedua, kolom ketiga:", matriks_nilai[1, 2])
print("Baris ketiga, kolom pertama:", matriks_nilai[2, 0])
```

### Sub Konteks 09: Memilih Baris dan Kolom
`a[i, :]` memilih satu baris lengkap; `a[:, j]` memilih satu kolom lengkap. Tanda `:` berarti mengambil semua indeks di sepanjang sumbu tersebut.

```python
# Memilih satu baris dan satu kolom dari matriks 2D
print("Matriks nilai:\n", matriks_nilai)
print("Baris kedua:", matriks_nilai[1, :])
print("Kolom ketiga:", matriks_nilai[:, 2])
print("Rata-rata kolom kedua:", np.mean(matriks_nilai[:, 1]))
```

### Sub Konteks 10: Slicing 1D, 2D, dan `step`
Slicing memakai bentuk `a[start:stop:step]`. Indeks `start` ikut, sedangkan `stop` tidak ikut. Untuk 2D gunakan dua rentang, misalnya `a[1:3, 0:2]`. Bentuk `a[::2]` mengambil setiap dua posisi dan `a[::-1]` membalik urutan. Waspadai kesalahan indeks di luar batas.

```python
# Slicing 1D, 2D, dan langkah
urutan = np.array([10, 20, 30, 40, 50, 60])
grid = np.arange(1, 26).reshape(5, 5)

print("Index 1 s.d. 3:", urutan[1:4]) #dari indeks ke 1 hingga indeks 3
print("Tiga pertama:", urutan[:3])
print("Tiga terakhir:", urutan[-3:])
print("Setiap dua posisi:", urutan[::2])
print("Urutan terbalik:", urutan[::-1])
print("Submatriks baris 2-4, kolom 3-5:\n", grid[1:4, 2:5])
print("Setiap dua baris dan kolom:\n", grid[::2, ::2])

try:
    print(urutan[6])  # Salah: indeks maksimum urutan adalah 5
except IndexError as error:
    print("Contoh indeks melewati batas:", type(error).__name__)
```
