**Tujuan:** mahasiswa mampu mengurutkan dan memilih data berdasarkan posisi maupun kondisi, memperbarui isi array, menggabungkan dan menghapus bagian array, menghitung statistik ringkas, serta menerapkan operasi matematis secara vektorisasi menggunakan NumPy.

### Sub Konteks 01: Mengurutkan Array 1D dengan `np.sort()`

**Sorting** adalah proses mengurutkan elemen array. Fungsi `np.sort()` menghasilkan array baru yang telah terurut, sehingga array asal tidak berubah. Secara default, urutan yang dihasilkan adalah menaik (*ascending*). Untuk memperoleh urutan menurun (*descending*), gunakan `[::-1]` pada hasil pengurutan.

```python
# Mengimpor NumPy
import numpy as np

# Data nilai akhir lima mahasiswa
nilai_akhir = np.array([78, 92, 85, 68, 90])

nilai_menaik = np.sort(nilai_akhir)
nilai_menurun = np.sort(nilai_akhir)[::-1]

print("Data asli       :", nilai_akhir)
print("Urutan menaik   :", nilai_menaik)
print("Urutan menurun  :", nilai_menurun)
```

### Sub Konteks 02: Mengurutkan Array 2D, `axis`, dan `np.argsort()`

Pada array 2D, `np.sort()` mengurutkan elemen berdasarkan sumbu (*axis*) yang dipilih.

- `axis=0`: mengurutkan nilai dari atas ke bawah pada setiap kolom.
- `axis=1`: mengurutkan nilai dari kiri ke kanan pada setiap baris.
- `axis=None`: mengurutkan seluruh elemen dan menghasilkan array 1D.

Fungsi `np.argsort()` menghasilkan **indeks urutan**, bukan nilai yang telah diurutkan. Indeks tersebut berguna ketika data nilai harus tetap berpasangan dengan identitas mahasiswa.

```python
# Baris = mahasiswa; kolom = komponen penilaian
matriks_nilai = np.array([
    [78, 82, 80],
    [92, 88, 95],
    [65, 74, 70],
    [85, 90, 87],
    [90, 86, 92]
])

print("Matriks nilai:\n", matriks_nilai)
print("Urut setiap kolom (axis=0):\n", np.sort(matriks_nilai, axis=0))
print("Urut setiap baris (axis=1):\n", np.sort(matriks_nilai, axis=1))
print("Urut seluruh elemen:\n", np.sort(matriks_nilai, axis=None))

# Mengurutkan nama mahasiswa berdasarkan nilai akhir
nama_mahasiswa = np.array(["Alya", "Bima", "Citra", "Daffa", "Eka"])
nilai_mahasiswa = np.array([78, 92, 85, 68, 90])

indeks_urut = np.argsort(nilai_mahasiswa)
print("Indeks urutan nilai:", indeks_urut)
print("Nama berdasarkan nilai menaik:", nama_mahasiswa[indeks_urut])
print("Nilai berdasarkan urutan:", nilai_mahasiswa[indeks_urut])
```

### Sub Konteks 03: Boolean Mask dan Boolean Indexing

**Boolean mask** adalah array yang berisi `True` dan `False` sebagai hasil operasi perbandingan. Boolean indexing menggunakan mask tersebut untuk mengambil elemen yang memenuhi kondisi.

Misalnya, kondisi `nilai >= 80` menghasilkan mask. Ketika mask diterapkan sebagai `nilai[mask]`, hanya nilai yang memenuhi kondisi yang akan ditampilkan.

```python
# Data nilai ujian mahasiswa
nilai_ujian = np.array([65, 78, 84, 92, 71, 88, 95, 74])

mask_lulus_target = nilai_ujian >= 80

print("Nilai ujian:", nilai_ujian)
print("Boolean mask nilai >= 80:", mask_lulus_target)
print("Nilai minimal 80:", nilai_ujian[mask_lulus_target])
print("Nilai di bawah 80:", nilai_ujian[nilai_ujian < 80])

# Memilih nilai pada rentang 75 sampai 90, termasuk batasnya
nilai_rentang = nilai_ujian[
    (nilai_ujian >= 75) & (nilai_ujian <= 90)
]
print("Nilai pada rentang 75-90:", nilai_rentang)
```

### Sub Konteks 04: Filtering Array 2D, Fancy Indexing, dan `np.where()`

Pada array 2D, Boolean indexing dapat digunakan untuk memilih **baris** berdasarkan kondisi pada kolom tertentu. Sementara itu, **Fancy indexing** memilih elemen atau baris berdasarkan daftar indeks posisi yang ditentukan.

`np.where()` memiliki dua bentuk penggunaan utama:

1. `np.where(kondisi)` menghasilkan indeks posisi yang memenuhi kondisi.
2. `np.where(kondisi, nilai_jika_benar, nilai_jika_salah)` menghasilkan nilai berdasarkan kondisi.

Untuk menggabungkan beberapa kondisi, gunakan `&` untuk AND, `|` untuk OR, dan `~` untuk NOT. Setiap kondisi perlu diberi tanda kurung. Jangan menggunakan `and` atau `or` untuk membandingkan seluruh array NumPy.

```python
# Memilih mahasiswa dengan nilai komponen pertama minimal 80
baris_terpilih = matriks_nilai[matriks_nilai[:, 0] >= 80]

print("Baris dengan nilai komponen pertama >= 80:\n", baris_terpilih)
print("Shape hasil pilihan:", baris_terpilih.shape)

# Fancy indexing: mengambil elemen pada indeks 0, 2, dan 4
data_contoh = np.array([10, 20, 30, 40, 50])
print("Fancy indexing:", data_contoh[[0, 2, 4]])

# np.where(kondisi): memperoleh indeks
curah_hujan = np.array([0, 12, 8, 25, 4, 18])
indeks_hujan = np.where(curah_hujan >= 10)[0]
print("Indeks curah hujan >= 10:", indeks_hujan)
print("Nilai curah hujan >= 10:", curah_hujan[indeks_hujan])

# np.where(kondisi, benar, salah): menentukan label
status_hujan = np.where(curah_hujan > 0, "Hujan", "Tidak hujan")
print("Status hujan:", status_hujan)

# Menggunakan beberapa kondisi
suhu = np.array([26, 29, 32, 35, 28, 31])
suhu_normal = suhu[(suhu >= 28) & (suhu <= 32)]
suhu_ekstrem = suhu[(suhu < 28) | (suhu > 32)]

print("Suhu 28-32:", suhu_normal)
print("Suhu di bawah 28 atau di atas 32:", suhu_ekstrem)
```

### Sub Konteks 05: Memperbarui Elemen, Baris, dan Kolom Array

Nilai array dapat diperbarui melalui assignment menggunakan indeks. Pada array 2D, `array[baris, kolom]` memilih satu elemen, `array[baris, :]` memilih satu baris, dan `array[:, kolom]` memilih satu kolom.

Gunakan `.copy()` sebelum melakukan perubahan apabila data asli harus dipertahankan. Tanpa `.copy()`, assignment variabel baru dapat tetap merujuk pada objek array yang sama jika hanya menggunakan penugasan biasa (`data_baru = data_asli`).

```python
# Memperbarui satu elemen tanpa mengubah data asli
data_asli = np.array([10, 20, 30, 40])
data_revisi = data_asli.copy()
data_revisi[1] = 25

print("Data asli:", data_asli)
print("Data revisi:", data_revisi)

# Memperbarui elemen, satu baris, dan satu kolom
nilai_kelas = np.array([
    [70, 75, 80],
    [82, 86, 88],
    [90, 91, 93]
])

nilai_kelas[0, 0] = 78
nilai_kelas[1, :] = [84, 87, 90]
nilai_kelas[:, 2] = 95

print("Matriks setelah diperbarui:\n", nilai_kelas)
```

### Sub Konteks 06: Conditional Replacement, Menambah, dan Menghapus Data

**Conditional replacement** mengganti nilai yang memenuhi kondisi tertentu. Penggantian dapat dilakukan dengan Boolean indexing atau `np.where()`.

`np.concatenate()` menggabungkan array:
- `axis=0` menambahkan baris. Jumlah kolom harus sama.
- `axis=1` menambahkan kolom. Jumlah baris harus sama.

`np.delete()` menghapus elemen, baris, atau kolom dan menghasilkan array baru. Gunakan `axis=0` untuk menghapus baris dan `axis=1` untuk menghapus kolom. Periksa `.shape` agar bentuk data tetap sesuai kebutuhan analisis.

```python
# Mengganti nilai negatif yang dianggap sebagai kode pencatatan tidak valid
pengukuran_asli = np.array([12, -2, 18, -5, 9])
pengukuran_bersih = pengukuran_asli.copy()
pengukuran_bersih[pengukuran_bersih < 0] = 0

print("Pengukuran asli:", pengukuran_asli)
print("Hasil Boolean indexing:", pengukuran_bersih)
print("Hasil np.where:", np.where(pengukuran_asli < 0, 0, pengukuran_asli))

# Menambahkan baris baru
penjualan = np.array([
    [12, 15],
    [10, 18]
])
baris_baru = np.array([[14, 16]])

penjualan_tambah_baris = np.concatenate(
    (penjualan, baris_baru), axis=0
)

print("Setelah menambah baris:\n", penjualan_tambah_baris)
print("Shape:", penjualan_tambah_baris.shape)

# Menambahkan kolom baru; kolom harus berbentuk 2D (jumlah_baris, 1)
kolom_baru = np.array([[20], [21], [22]])
penjualan_lengkap = np.concatenate(
    (penjualan_tambah_baris, kolom_baru), axis=1
)

print("Setelah menambah kolom:\n", penjualan_lengkap)
print("Shape:", penjualan_lengkap.shape)

# Menghapus baris kedua dan kolom ketiga
tanpa_baris = np.delete(penjualan_lengkap, 1, axis=0)
tanpa_kolom = np.delete(penjualan_lengkap, 2, axis=1)

print("Setelah menghapus baris kedua:\n", tanpa_baris)
print("Setelah menghapus kolom ketiga:\n", tanpa_kolom)
```

### Sub Konteks 07: Kompatibilitas Shape dan Dimension saat Menggabungkan Array

Array yang digabungkan harus memiliki jumlah dimensi yang sama. Untuk `axis=0`, jumlah kolom harus sama; untuk `axis=1`, jumlah baris harus sama. Vektor 1D perlu diubah menjadi baris atau kolom 2D menggunakan `reshape()` jika diperlukan.

Kesalahan umum adalah menggabungkan matriks berukuran `(2, 3)` dengan vektor berukuran `(2,)` menggunakan `axis=1`. Bentuk vektor tersebut perlu diubah menjadi `(2, 1)` terlebih dahulu.

```python
# Matriks berukuran 2 x 3
matriks = np.arange(1, 7).reshape(2, 3)
# 1 2 3
# 4 5 6

# Menyiapkan baris baru (1 x 3) dan kolom baru (2 x 1)
nilai_baris = np.array([7, 8, 9]).reshape(1, 3)
nilai_kolom = np.array([10, 11]).reshape(2, 1)

print("Shape matriks:", matriks.shape)
print("Shape baris baru:", nilai_baris.shape)
print("Shape kolom baru:", nilai_kolom.shape)

print("Gabungan baris:\n", np.concatenate((matriks, nilai_baris), axis=0))
print("Gabungan kolom:\n", np.concatenate((matriks, nilai_kolom), axis=1))

# Contoh kesalahan shape dan cara memperbaikinya
try:
    np.concatenate((matriks, np.array([10, 11])), axis=1)
except ValueError as error:
    print("Bentuk tidak kompatibel:", type(error).__name__)
```

### Sub Konteks 08: Array Aggregation dengan `sum`, `min`, `max`, dan `mean`

**Aggregation** merangkum sekumpulan nilai menjadi satu atau beberapa ukuran statistik. Fungsi yang sering digunakan adalah:

- `np.sum()`: jumlah seluruh nilai.
- `np.min()`: nilai minimum.
- `np.max()`: nilai maksimum.
- `np.mean()`: nilai rata-rata.

Pada array 2D, `axis=0` menghasilkan ringkasan per kolom, sedangkan `axis=1` menghasilkan ringkasan per baris. Jika baris mewakili mahasiswa dan kolom mewakili ujian, `axis=0` menghasilkan statistik per ujian dan `axis=1` menghasilkan statistik per mahasiswa.

```python
# Jumlah pengunjung selama lima hari
pengunjung_harian = np.array([120, 135, 128, 142, 150])

print("Total pengunjung:", np.sum(pengunjung_harian))
print("Pengunjung minimum:", np.min(pengunjung_harian))
print("Pengunjung maksimum:", np.max(pengunjung_harian))
print("Rata-rata pengunjung:", np.mean(pengunjung_harian))

# Baris = mahasiswa = 1; kolom = ujian = 0
nilai_ujian = np.array([
    [80, 85, 90],
    [70, 75, 80],
    [90, 92, 94]
])

print("Total per ujian:", np.sum(nilai_ujian, axis=0))
print("Rata-rata per mahasiswa:", np.mean(nilai_ujian, axis=1))
print("Maksimum per ujian:", np.max(nilai_ujian, axis=0))
```

### Sub Konteks 09: Cumulative Sum dan `keepdims`

`np.cumsum()` menghitung jumlah kumulatif secara bertahap. Pada array 1D, hasil pada setiap posisi merupakan jumlah nilai dari awal hingga posisi tersebut. Pada array 2D, `axis` menentukan arah perhitungan kumulatif.

Secara default, fungsi agregasi menghilangkan sumbu yang diringkas. Argumen `keepdims=True` mempertahankan sumbu tersebut dengan ukuran 1. Hasil ini berguna untuk operasi lanjutan yang membutuhkan bentuk array kompatibel.

```python
# Jumlah penjualan selama empat minggu
penjualan_mingguan = np.array([10, 15, 12, 18])

print("Penjualan mingguan:", penjualan_mingguan)
print("Penjualan kumulatif:", np.cumsum(penjualan_mingguan))

#nilai_ujian = np.array([
#    [80, 85, 90],
#    [70, 75, 80],
#    [90, 92, 94]
#])
# Kumulatif nilai setiap mahasiswa dari ujian pertama sampai ketiga
print("Kumulatif nilai per mahasiswa:\n", np.cumsum(nilai_ujian, axis=1))

# Membandingkan hasil mean tanpa dan dengan keepdims
rata_mahasiswa = np.mean(nilai_ujian, axis=1)
rata_mahasiswa_2d = np.mean(nilai_ujian, axis=1, keepdims=True)

print("Mean tanpa keepdims:", rata_mahasiswa)
print("Shape tanpa keepdims:", rata_mahasiswa.shape)
print("Mean dengan keepdims:\n", rata_mahasiswa_2d)
print("Shape dengan keepdims:", rata_mahasiswa_2d.shape)
```

### Sub Konteks 10: Vectorization, Operasi Element-wise, dan Perbandingan

**Vectorization** memungkinkan operasi dilakukan langsung pada seluruh elemen array tanpa menulis perulangan untuk setiap elemen. Cara ini membuat syntax lebih ringkas dan umumnya efisien untuk komputasi numerik.

Operasi skalar seperti `+`, `-`, `*`, `/`, dan `**` diterapkan pada setiap elemen array. Operasi antara dua array dengan bentuk kompatibel dilakukan secara **element-wise**, yaitu pada elemen yang posisinya bersesuaian.

Operator perbandingan (`>`, `<`, `>=`, `<=`, `==`, dan `!=`) menghasilkan array Boolean yang dapat digunakan untuk filtering.

```python
# Perbandingan Python loop dan NumPy vectorization
data = np.array([2, 4, 6, 8])

hasil_loop = []
for nilai in data:
    hasil_loop.append(nilai * 2)

hasil_vector = data * 2

print("Hasil Python loop:", hasil_loop)
print("Hasil vectorization:", hasil_vector)

# Operasi aritmetika skalar
nilai = np.array([10, 20, 30])

print("Tambah 5:", nilai + 5)
print("Kurang 2:", nilai - 2)
print("Kali 3:", nilai * 3)
print("Bagi 2:", nilai / 2)
print("Kuadrat:", nilai ** 2)

# Operasi element-wise antara dua array
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Penjumlahan:", a + b)
print("Pengurangan:", a - b)
print("Perkalian element-wise:", a * b)
print("Pembagian:", a / b)

# Operasi perbandingan menghasilkan Boolean array
nilai_aktual = np.array([75, 90, 82, 60])
nilai_target = np.array([80, 85, 80, 70])

print("Aktual >= target:", nilai_aktual >= nilai_target)
print("Aktual == target:", nilai_aktual == nilai_target)
print("Nilai yang mencapai target:", nilai_aktual[nilai_aktual >= nilai_target])
```

### Sub Konteks 11: Broadcasting dan Penerapan pada Data Statistik

**Broadcasting** memungkinkan operasi antara array dengan bentuk berbeda selama dimensinya kompatibel. NumPy membandingkan ukuran dimensi dari sisi paling kanan; ukuran yang bersesuaian harus sama atau salah satunya bernilai 1.

Contoh berikut menambahkan vektor koreksi berisi tiga nilai ke setiap baris matriks yang memiliki tiga kolom. Contoh berikutnya menghitung deviasi setiap pengamatan terhadap rata-ratanya tanpa menggunakan perulangan.

Perlu dibedakan bahwa operator `*` pada dua array melakukan perkalian element-wise, bukan perkalian matriks. Perkalian matriks menggunakan operator `@` atau fungsi `np.matmul()`.

```python
# Broadcasting: koreksi diterapkan pada setiap baris
matriks_nilai = np.array([
    [70, 75, 80],
    [82, 86, 88],
    [90, 91, 93]
])
koreksi_per_ujian = np.array([1, 2, 3])

hasil_broadcast = matriks_nilai + koreksi_per_ujian

print("Shape matriks:", matriks_nilai.shape)
print("Shape koreksi:", koreksi_per_ujian.shape)
print("Hasil broadcasting:\n", hasil_broadcast)

# Perbedaan perkalian element-wise dan perkalian matriks
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

#A 1 2
#  3 4

#B 5 6
#  7 8
print("Perkalian element-wise:\n", A * B)
print("Perkalian matriks:\n", A @ B)

# Menghitung deviasi pengamatan terhadap rata-rata
observasi = np.array([12.0, 15.0, 13.0, 20.0]) # (i = 1,..,4)
rata_observasi = np.mean(observasi)
deviasi = observasi - rata_observasi
#Xi-Xbar

print("Observasi:", observasi)
print("Rata-rata:", rata_observasi)
print("Deviasi dari rata-rata:", deviasi)
print("Jumlah deviasi (mendekati nol):", np.sum(deviasi))
```

```python
# Lengkapi sintaks yang rumpang
# Nama : Muhammad Mirza Apriansyah
# NIM  : 2511017310004

import numpy as np

# A. Memperbarui data tanpa mengubah array asli
data_asli = np.array([45, -2, 52, -1, 48])
data_bersih = data_asli.copy()
data_bersih[data_bersih < 0] = 0

print("Data asli:", data_asli)
print("Data bersih:", data_bersih)

# B. Memperbarui elemen, baris, dan kolom
nilai_kelas = np.array([
    [70, 75, 80],
    [82, 86, 88],
    [90, 91, 93]
])

nilai_kelas[0, 1] = 80
nilai_kelas[1, :] = [84, 87, 90]
nilai_kelas[:, 2] = 95

print("Nilai setelah pembaruan:\n", nilai_kelas)

# C. Conditional replacement menggunakan np.where()
nilai_koreksi = np.where(nilai_kelas < 75, 75, nilai_kelas)
print("Nilai setelah batas minimum:\n", nilai_koreksi)

# D. Menambahkan baris dan kolom
baris_baru = np.array([[84, 87, 89]])
kelas_tambah_baris = np.concatenate(
    (nilai_koreksi, baris_baru), axis=0
)

kolom_baru = np.array([80, 85, 90, 95]).reshape(4, 1)
kelas_lengkap = np.concatenate(
    (kelas_tambah_baris, kolom_baru), axis=1
)

print("Setelah menambah baris dan kolom:\n", kelas_lengkap)
print("Shape akhir:", kelas_lengkap.shape)

# E. Menghapus baris dan kolom
hapus_baris = np.delete(kelas_lengkap, 0, axis=0)
hapus_kolom = np.delete(kelas_lengkap, -1, axis=1)

print("Tanpa baris pertama:\n", hapus_baris)
print("Tanpa kolom terakhir:\n", hapus_kolom)

# F. Aggregation pada data pengunjung
pengunjung = np.array([
    [20, 25, 30],
    [18, 22, 28],
    [24, 27, 32]
])

total_seluruh = np.sum(pengunjung)
minimum = np.min(pengunjung)
maksimum = np.max(pengunjung)
rata_rata = np.mean(pengunjung)

total_per_lokasi = np.sum(pengunjung, axis=0)
rata_per_hari = np.mean(pengunjung, axis=1)

print("Total:", total_seluruh)
print("Minimum dan maksimum:", minimum, maksimum)
print("Rata-rata keseluruhan:", rata_rata)
print("Total per lokasi:", total_per_lokasi)
print("Rata-rata per hari:", rata_per_hari)

# G. Cumulative sum dan keepdims
kumulatif = np.cumsum(pengunjung, axis=1)
rata_per_hari_2d = np.mean(
    pengunjung, axis=1, keepdims=True
)

print("Kumulatif per baris:\n", kumulatif)
print("Mean per hari dengan keepdims:\n", rata_per_hari_2d)
print("Shape mean:", rata_per_hari_2d.shape)

# H. Vectorization dan operasi element-wise
harga = np.array([10000, 15000, 20000, 25000])
jumlah = np.array([2, 3, 1, 4])

pendapatan = harga * jumlah
harga_diskon = harga * 0.9
mask_pendapatan = pendapatan >= 40000

print("Pendapatan per produk:", pendapatan)
print("Harga setelah diskon:", harga_diskon)
print("Mask pendapatan:", mask_pendapatan)

# I. Broadcasting: koreksi untuk setiap kolom
nilai_ujian = np.array([
    [70, 75, 80],
    [80, 85, 90],
    [90, 92, 95]
])
koreksi = np.array([2, 3, 1])
hasil_koreksi = nilai_ujian + koreksi

print("Hasil broadcasting:\n", hasil_koreksi)

# J. Menghitung deviasi setiap pengamatan dari rata-rata
observasi = np.array([12.0, 15.0, 13.0, 20.0])
mean_observasi = np.mean(observasi)
deviasi = observasi - mean_observasi

print("Mean:", mean_observasi)
print("Deviasi:", deviasi)
print("Jumlah deviasi:", np.sum(deviasi))
```
